```python
import pandas as pd
import glob
```

```python
path_stamps = glob.glob('/home/dmoreno/pipeline_v4_final/pipeline/training/stamp_classifier/data_acquisition/rubin/data/processed/ts_stamps_v0.0.2_comm_4candmax/stamps/*')
path_stamps
```

```python
stamps = []
for path in path_stamps:
    stamps.append(pd.read_pickle(path))

stamps = pd.concat(stamps)
stamps
```

```python
test_predictions = pd.read_csv('test_predictions.csv')
oid_mid = (test_predictions.head(10).oid.values, test_predictions.head(10).measurement_id.values)
oid_mid
```

```python
calibrated = pd.read_parquet('/home/dmoreno/pipeline_v4_final/pipeline/training/stamp_classifier/models/experimentation/stamp_rubin/inference_output/2.0.0_eval/predictions_test.parquet')
calibrated
```

```python
calibrated = calibrated[calibrated.oid.isin(test_predictions.head(10).oid.values)]
calibrated
```

```python
rubin_stamp_test = stamps[
    (stamps.oid.isin(test_predictions.head(10).oid.values)) &
    (stamps.measurement_id.isin(test_predictions.head(10).measurement_id.values))
    ]

rubin_stamp_test
```

```python
rubin_stamp_test.columns
```

```python
comm_objs_4candmax = pd.read_parquet('/home/dmoreno/pipeline_v4_final/pipeline/training/stamp_classifier/data_acquisition/rubin/data/processed/ts_stamps_v0.0.2_comm_4candmax/comm_objs_4candmax.parquet')
comm_objs_4candmax
```

```python
resultado = pd.merge(
    rubin_stamp_test, 
    comm_objs_4candmax, 
    on=['oid', 'measurement_id'], 
    how='left'
)
resultado
```

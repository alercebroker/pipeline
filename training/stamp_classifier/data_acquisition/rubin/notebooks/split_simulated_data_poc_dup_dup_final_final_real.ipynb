```python
import pandas as pd
import numpy as np
from sklearn.model_selection import StratifiedKFold

unique_objects_df = df_real_clean[['oid_kept', 'class']].drop_duplicates().reset_index(drop=True)

# 2. División en Test y Train/Val
test_oids = []
trainval_oids = []
for class_name in unique_objects_df['class'].unique():
    class_oids_df = unique_objects_df[unique_objects_df['class'] == class_name]
    shuffled_oids = class_oids_df['oid_kept'].sample(frac=1, random_state=42)
    test_frac = 0.60 if class_name == 'SN' else 0.20
    split_idx = int(len(shuffled_oids) * test_frac)
    test_oids.extend(shuffled_oids.iloc[:split_idx])
    trainval_oids.extend(shuffled_oids.iloc[split_idx:])
    
test_objects_df = unique_objects_df[unique_objects_df['oid_kept'].isin(test_oids)]
trainval_objects_df = unique_objects_df[unique_objects_df['oid_kept'].isin(trainval_oids)]

# 3. Crear los 5 Folds con la lógica personalizada
N_SPLITS = 5
all_partitions = []

df_test_partitions = test_objects_df.copy()
df_test_partitions['partition'] = 'test'
all_partitions.append(df_test_partitions)

sn_trainval_df = trainval_objects_df[trainval_objects_df['class'] == 'SN']
other_trainval_df = trainval_objects_df[trainval_objects_df['class'] != 'SN']

if not other_trainval_df.empty:
    skf = StratifiedKFold(n_splits=N_SPLITS, shuffle=True, random_state=42)
    other_trainval_df['fold'] = -1
    for fold_num, (train_idx, val_idx) in enumerate(skf.split(other_trainval_df, other_trainval_df['class'])):
        val_oids = other_trainval_df.iloc[val_idx]['oid_kept']
        other_trainval_df.loc[other_trainval_df['oid_kept'].isin(val_oids), 'fold'] = fold_num

for i in range(N_SPLITS):
    if not other_trainval_df.empty:
        df_other_perspective = other_trainval_df.copy()
        df_other_perspective['partition'] = np.where(df_other_perspective['fold'] == i, f'validation_{i}', f'training_{i}')
        all_partitions.append(df_other_perspective[['oid_kept', 'class', 'partition']])

    if not sn_trainval_df.empty:
        validation_sn_oids = sn_trainval_df['oid_kept'].sample(frac=0.6, random_state=42 + i)
        df_sn_perspective = sn_trainval_df.copy()
        df_sn_perspective['partition'] = np.where(df_sn_perspective['oid_kept'].isin(validation_sn_oids), f'validation_{i}', f'training_{i}')
        all_partitions.append(df_sn_perspective[['oid_kept', 'class', 'partition']])

main_mapping_df = pd.concat(all_partitions, ignore_index=True)

# Unir el mapeo con el DataFrame original principal
final_partition_df_main = pd.merge(
    df_real_clean,
    main_mapping_df,
    on=['oid_kept', 'class'],
    how='left'
)

# --- PASO B: Procesar el DataFrame adicional (df_dup_tot) ---
print("\n--- Procesando datos adicionales para entrenamiento ---")

extra_training_partitions = []
# Replicamos los datos para cada uno de los 5 folds, asignándolos siempre a 'training'
for i in range(N_SPLITS):
    df_fold_extra = df_dup_tot.copy()
    df_fold_extra['partition'] = f'training_{i}'
    extra_training_partitions.append(df_fold_extra)

# Concatenamos las 5 réplicas en un único DataFrame
df_extra_partitioned = pd.concat(extra_training_partitions, ignore_index=True)

print(f"Se han generado {len(df_extra_partitioned)} filas a partir de los datos adicionales.")
print(f"({len(df_dup_tot)} filas originales x {N_SPLITS} folds)")
print("-" * 50)


# --- PASO C: Combinar ambos resultados ---

final_partition_df = pd.concat(
    [final_partition_df_main, df_extra_partitioned],
    ignore_index=True
)


# --- 4. Verificación y Formato de Salida ---
print("\n--- Verificación Final (Fold 0) ---")
# Tomamos un objeto del set adicional para verificar
oid_to_check = df_dup_tot['oid_kept'].iloc[0]
print(f"Verificando particiones para el objeto de muestra '{oid_to_check}':")
print(final_partition_df[final_partition_df['oid_kept'] == oid_to_check][['oid_kept', 'partition']])

print("\nConteo de objetos únicos en las particiones del Fold 0:")
check_df = final_partition_df[['oid_kept', 'class', 'partition']].drop_duplicates()
fold_0_counts = check_df[check_df['partition'].isin(['training_0', 'validation_0'])]
print(fold_0_counts.groupby(['partition', 'class']).size().unstack(fill_value=0))
print("-" * 50)


# Reordenar las columnas para el formato final
output_columns = [
    'oid', 'oid_kept', 'measurement_id', 'band', 'ra', 'dec', 'mjd', 
    'class', 'partition', 'dataset_origin', 'synth_file_path'
]
final_partition_df = final_partition_df[output_columns]
```

```python
print("\n--- INICIANDO VERIFICACIÓN PRINCIPAL ---")
print("Para cada objeto (oid), se comprobará que todas sus observaciones (measurement_id) tienen el mismo conjunto de particiones.\n")

# Agrupamos por el identificador del objeto
grouped_by_object = final_dataset_partitioned.groupby('oid')

inconsistent_objects = []

for oid_name, group in grouped_by_object:
    # Obtenemos la lista de observaciones únicas para este objeto
    unique_measurements = group['measurement_id'].unique()
    
    # Si solo hay una observación, es consistente por definición
    if len(unique_measurements) <= 1:
        continue
        
    # Tomamos el conjunto de particiones de la primera observación como referencia
    first_measurement_partitions = set(
        group[group['measurement_id'] == unique_measurements[0]]['partition']
    )
    
    # Comparamos el conjunto de particiones de las demás observaciones con la referencia
    for mid in unique_measurements[1:]:
        current_measurement_partitions = set(
            group[group['measurement_id'] == mid]['partition']
        )
        
        # Si los conjuntos no son idénticos, hemos encontrado una inconsistencia
        if current_measurement_partitions != first_measurement_partitions:
            inconsistent_objects.append(oid_name)
            print(f"🔴 INCONSISTENCIA ENCONTRADA en el objeto: '{oid_name}'")
            print(f"   - Particiones de la observación '{unique_measurements[0]}': {first_measurement_partitions}")
            print(f"   - Particiones de la observación '{mid}': {current_measurement_partitions}\n")
            # Rompemos el bucle interior porque ya sabemos que este objeto es inconsistente
            break

# --- Reporte Final de la Verificación Principal ---
print("\n--- REPORTE DE VERIFICACIÓN PRINCIPAL ---")
if not inconsistent_objects:
    print("✅ ¡Éxito! Todas las observaciones de cada objeto están consistentemente en los mismos subsets (test/training/validation).")
else:
    print(f"❌ ¡Error! Se encontraron inconsistencias en los siguientes objetos: {inconsistent_objects}")
print("-" * 50)
```

```python
# --------------------------------------------------------

# Nombres de tus columnas
OID_COL = 'oid'  # Columna con el ID del objeto
RA_COL = 'ra'    # Columna de Ascensión Recta
DEC_COL = 'dec'  # Columna de Declinación

print(f"Total de objetos: {len(df_real_clean)}")

# ==============================================================================
# 2. CALCULAR DISTANCIAS AL VECINO MÁS CERCANO
# ==============================================================================

# Extraemos los valores como arrays de numpy float para evitar errores de PyArrow
ra_vals = df_real_clean[RA_COL].astype(float).to_numpy()
dec_vals = df_real_clean[DEC_COL].astype(float).to_numpy()

# Crear objeto de coordenadas de Astropy
coords = SkyCoord(ra=ra_vals*u.degree, dec=dec_vals*u.degree)

# Buscar para cada objeto, cuál es su vecino más cercano en el MISMO catálogo.
# Usamos nthneighbor=2 porque el vecino #1 es el objeto mismo (distancia 0).
idx, d2d, _ = coords.match_to_catalog_sky(coords, nthneighbor=2)

# ==============================================================================
# 3. CREAR DATAFRAME DE RESULTADOS
# ==============================================================================
# 'idx' son los índices de los vecinos en el DataFrame original
# 'd2d' son las distancias angulares

resultados = pd.DataFrame({
    'oid_1': df_real_clean[OID_COL].values,           # Objeto original
    'oid_2': df_real_clean.iloc[idx][OID_COL].values, # Su vecino más cercano
    'distancia_arcsec': d2d.arcsec         # Distancia en arcosegundos
})

# Ordenar por distancia (las más pequeñas primero)
resultados_ordenados = resultados.sort_values('distancia_arcsec')

# ==============================================================================
# 4. MOSTRAR LOS PARES MÁS CERCANOS
# ==============================================================================
print("\n--- Top 10 Pares de Objetos Más Cercanos ---")
print(resultados_ordenados.head(10))

# Opcional: Filtrar objetos extremadamente cercanos (posibles duplicados)
limite = 4.0 # arcsec
muy_cercanos = resultados_ordenados[resultados_ordenados['distancia_arcsec'] < limite]

if not muy_cercanos.empty:
    print(f"\n⚠️ Se encontraron {len(muy_cercanos)} pares con distancia < {limite} arcsec.")
else:
    print(f"\n✅ No se encontraron objetos con distancia < {limite} arcsec.")
```

```python
# PREPARACIÓN PARA StratifiedKFold
from sklearn.model_selection import StratifiedKFold

X = df_train_val_pool.index.to_numpy()
y = df_train_val_pool[CLASS_COL].to_numpy()

print(f"X shape: {X.shape}, y shape: {y.shape}")
print(f"Distribución de clases: {pd.Series(y).value_counts()}")

# EJECUCIÓN DE StratifiedKFold
skf = StratifiedKFold(
    n_splits=N_FOLDS, 
    shuffle=True, 
    random_state=RANDOM_STATE
)

# Inicializar columna fold
df_train_val_pool = df_train_val_pool.copy()
df_train_val_pool['fold'] = -1

# Iterar sobre los splits
for fold_idx, (train_idx, val_idx) in enumerate(skf.split(X, y)):
    print(f"Procesando fold {fold_idx}: {len(train_idx)} train, {len(val_idx)} val")
    df_train_val_pool.loc[df_train_val_pool.index[val_idx], 'fold'] = fold_idx

# Verificar asignación completa
unassigned = (df_train_val_pool['fold'] == -1).sum()
if unassigned > 0:
    print(f"ADVERTENCIA: {unassigned} filas no recibieron asignación de fold.")
    df_train_val_pool.loc[df_train_val_pool['fold'] == -1, 'fold'] = 0

# Asegurar que la columna 'dataset_origin' existe en df_train_val_pool
if 'dataset_origin' not in df_train_val_pool.columns:
    df_train_val_pool['dataset_origin'] = 'mixed'

# PREPARAR LOS DATOS ADICIONALES
# Asegurar que df_dup_tot_non_sn tiene las columnas necesarias
if 'dataset_origin' not in df_dup_tot_non_sn.columns:
    df_dup_tot_non_sn['dataset_origin'] = 'real'

# DEFINIR COLUMNAS BASE (sin 'partition' que se agregará después)
base_cols = [FINAL_OID_COL, 'oid', FINAL_CANDID_COL, CLASS_COL, 'dataset_origin', 'synth_file_path']

# Verificar qué columnas están disponibles en ambos dataframes
available_in_dup = [col for col in base_cols if col in df_dup_tot_non_sn.columns]
available_in_train = [col for col in base_cols if col in df_train_val_pool.columns]

print(f"\nColumnas disponibles en df_dup_tot_non_sn: {available_in_dup}")
print(f"Columnas disponibles en df_train_val_pool: {available_in_train}")

# Usar solo las columnas que están en ambos
common_cols = list(set(available_in_dup) & set(available_in_train))
print(f"Columnas comunes: {common_cols}")

print(f"\nDatos adicionales a agregar: {df_dup_tot_non_sn.shape}")
print(f"Distribución en df_dup_tot_non_sn: {df_dup_tot_non_sn[CLASS_COL].value_counts()}")

# Construir lista final de DataFrames
cols_to_save = [FINAL_OID_COL, 'oid', FINAL_CANDID_COL, CLASS_COL, 'partition', 'dataset_origin'] 
partitions_list = [df_test[cols_to_save]]

for fold_idx in range(N_FOLDS):
    # Entrenamiento: Todo lo que NO es este fold
    train_mask = df_train_val_pool['fold'] != fold_idx
    val_mask = df_train_val_pool['fold'] == fold_idx
    
    train_part = df_train_val_pool[train_mask].copy()
    val_part = df_train_val_pool[val_mask].copy()
    
    # AGREGAR LOS DATOS ADICIONALES SOLO AL TRAIN - VERSIÓN CORREGIDA
    # 1. Preparar train_part (asignar partition primero)
    train_part_prepared = train_part[common_cols].copy()
    train_part_prepared['partition'] = f'training_{fold_idx}'
    
    # 2. Preparar datos adicionales
    df_dup_for_fold = df_dup_tot_non_sn[common_cols].copy()
    df_dup_for_fold['partition'] = f'training_{fold_idx}'
    
    # 3. Concatenar
    train_part_with_dup = pd.concat([train_part_prepared, df_dup_for_fold], ignore_index=True)
    
    # 4. Preparar validation
    val_part_prepared = val_part[common_cols].copy()
    val_part_prepared['partition'] = f'validation_{fold_idx}'
    
    partitions_list.extend([train_part_with_dup, val_part_prepared])

# DataFrame final consolidado
final_df = pd.concat(partitions_list, ignore_index=True)
final_df
```

```python
from sklearn.model_selection import train_test_split
from sklearn.model_selection import StratifiedKFold
import pandas as pd

# Mantener la misma división de no-SN que antes
df_real_sn = df_real_clean[df_real_clean['class'] == 'SN']
df_real_non_sn = df_real_clean[df_real_clean['class'] != 'SN']
print(f"Real SN: {df_real_sn.shape}")
print(f"Real non-SN: {df_real_non_sn.shape}")

FINAL_OID_COL = 'oid_kept'
FINAL_CANDID_COL = 'measurement_id'
CLASS_COL = 'class'
RANDOM_STATE = 42
TEST_SIZE_NON_SN = 0.2
N_FOLDS = 5

# 1. MANTENER LA MISMA DIVISIÓN DE NO-SN QUE ANTES
oids_for_trainval = final_df[
    (final_df.partition == 'training_0') |
    (final_df.partition == 'validation_0')
    ][FINAL_OID_COL].astype(int).values

oids_for_test = final_df[
    (final_df.partition == 'test')
    ][FINAL_OID_COL].astype(int).values

df_non_sn_for_trainval = df_real_non_sn[df_real_non_sn[FINAL_OID_COL].isin(oids_for_trainval)]
df_non_sn_for_test = df_real_non_sn[df_real_non_sn[FINAL_OID_COL].isin(oids_for_test)]

# 2. DIVIDIR LAS SUPERNOVAS REALES EN TRAIN+VAL vs TEST
unique_sn_oids = df_real_sn[FINAL_OID_COL].unique().astype(int)
oids_sn_for_trainval, oids_sn_for_test = train_test_split(
    unique_sn_oids,
    test_size=0.5,
    random_state=RANDOM_STATE
)

df_sn_for_trainval = df_real_sn[df_real_sn[FINAL_OID_COL].isin(oids_sn_for_trainval)]
df_sn_for_test = df_real_sn[df_real_sn[FINAL_OID_COL].isin(oids_sn_for_test)]


# 3. CREAR LOS CONJUNTOS FINALES (Ligeramente modificado para mayor claridad)
# Asignar origen a los datos para claridad en el dataframe final
df_sn_for_trainval = df_sn_for_trainval.copy()
df_sn_for_trainval['dataset_origin'] = 'real_sn'
df_non_sn_for_trainval = df_non_sn_for_trainval.copy()
df_non_sn_for_trainval['dataset_origin'] = 'real_non_sn'
df_sinteticos = df_sinteticos.copy()
df_sinteticos['dataset_origin'] = 'synthetic_sn'
if 'df_dup_tot' in locals() and not df_dup_tot.empty:
    df_dup_tot = df_dup_tot.copy()
    df_dup_tot['dataset_origin'] = 'real'

# Test: SN reales (subconjunto) + no-SN (mismo que antes)
df_test = pd.concat([df_sn_for_test, df_non_sn_for_test], ignore_index=True)
df_test['dataset_origin'] = 'real' # Origen genérico para el test

# El pool para la validación cruzada SIGUE CONTENIENDO SOLO DATOS REALES
df_train_val_pool_real = pd.concat([
    df_sn_for_trainval, 
    df_non_sn_for_trainval
], ignore_index=True).reset_index(drop=True)

print(f"\n=== ESTADÍSTICAS DE LA DIVISIÓN ===")
print(f"Pool de datos REALES para train/val: {df_train_val_pool_real.shape}")
print(f"Datos sintéticos (solo para train): {df_sinteticos.shape}")
print(f"Test: {df_test.shape}")


# =============================================================================
# INICIO DE LA SECCIÓN MODIFICADA (Reemplaza las secciones 4 y 5)
# =============================================================================

# 4. PREPARACIÓN PARA LA DIVISIÓN PERSONALIZADA (SOBRE DATOS REALES)
print(f"\n--- 4. Preparando división personalizada en {N_FOLDS} Folds ---")

# Separar el pool de datos REALES en SN y no-SN
df_real_sn_pool = df_train_val_pool_real[df_train_val_pool_real['class'] == 'SN']
df_real_non_sn_pool = df_train_val_pool_real[df_train_val_pool_real['class'] != 'SN']

print(f"Total de SN Reales en pool: {len(df_real_sn_pool)}")
print(f"Total de No-SN Reales en pool: {len(df_real_non_sn_pool)}")

# Para las no-SN, creamos 5 folds disjuntos (20% en validación)
skf_non_sn = StratifiedKFold(n_splits=N_FOLDS, shuffle=True, random_state=RANDOM_STATE)
non_sn_splits = list(skf_non_sn.split(df_real_non_sn_pool, df_real_non_sn_pool['class']))


# 5. CONSTRUIR DATAFRAME FINAL CON LÓGICA COMBINADA
# --- Verificación de columnas (lógica de seguridad)
# Definimos las columnas que queremos en el archivo final
desired_cols = [FINAL_OID_COL, 'oid', FINAL_CANDID_COL, CLASS_COL, 'dataset_origin', 'partition', 'synth_file_path']

# Encontramos las columnas comunes en los dataframes *de entrada*
all_dfs = [df_train_val_pool_real, df_test, df_sinteticos]
if 'df_dup_tot' in locals() and not df_dup_tot.empty:
    all_dfs.append(df_dup_tot)

common_cols_sets = [set(df.columns) for df in all_dfs]
available_cols = set.intersection(*common_cols_sets)

# <--- CAMBIO CLAVE 1: `cols_to_select` son las columnas comunes SIN 'dataset_origin' ni 'partition'
# que se añadirán manualmente. Esto es más seguro.
cols_to_select = [col for col in desired_cols if col in available_cols and col not in ['dataset_origin', 'partition']]
print(f"\nColumnas base comunes para seleccionar: {cols_to_select}")

# --- Construcción de particiones
partitions_list = []

# Añadir el conjunto de test
df_test_prepared = df_test.copy()
df_test_prepared['partition'] = 'test'
partitions_list.append(df_test_prepared)

# Bucle para crear cada uno de los 5 folds
for fold_idx in range(N_FOLDS):
    print(f"\n--- Creando Fold {fold_idx} ---")

    # 1. División de Supernovas Reales (50% validación)
    real_sn_train, real_sn_val = train_test_split(
        df_real_sn_pool, test_size=0.50, random_state=RANDOM_STATE + fold_idx
    )
    print(f"  SN Reales en validación: {len(real_sn_val)} ({len(real_sn_val)/len(df_real_sn_pool):.1%})")

    # 2. División de No-Supernovas Reales (20% validación, disjunto)
    train_idx_non_sn, val_idx_non_sn = non_sn_splits[fold_idx]
    non_sn_train = df_real_non_sn_pool.iloc[train_idx_non_sn]
    non_sn_val = df_real_non_sn_pool.iloc[val_idx_non_sn]
    print(f"  No-SN Reales en validación: {len(non_sn_val)} ({len(non_sn_val)/len(df_real_non_sn_pool):.1%})")

    # 3. Ensamblar los conjuntos
    val_part = pd.concat([real_sn_val, non_sn_val], ignore_index=True)
    val_part['partition'] = f'validation_{fold_idx}'
    
    train_part_real = pd.concat([real_sn_train, non_sn_train], ignore_index=True)
    
    train_part_final = pd.concat([train_part_real, df_sinteticos, df_dup_tot], ignore_index=True)
    train_part_final['partition'] = f'training_{fold_idx}'

    partitions_list.extend([train_part_final, val_part])

# DataFrame final consolidado
final_df_mixed_raw = pd.concat(partitions_list, ignore_index=True)

# <--- CAMBIO CLAVE 2: `cols_to_save` se construye a partir de las columnas base + las que añadimos.
# Esto garantiza que 'partition' y 'dataset_origin' estén presentes.
cols_to_save = cols_to_select + ['dataset_origin', 'partition']

# Filtramos el DataFrame final para quedarnos solo con las columnas que queremos y existen.
final_df_mixed = final_df_mixed_raw[[col for col in cols_to_save if col in final_df_mixed_raw.columns]]


# --- Verificación Final (Ahora debería funcionar) ---
print("\n=== VERIFICACIÓN FINAL ===")
print("Dimensiones del DataFrame final:", final_df_mixed.shape)
print("Columnas en el DataFrame final:", final_df_mixed.columns.tolist()) # Añadido para depuración

print("\nComposición de la partición 'validation_0':")
val_0_counts = final_df_mixed[final_df_mixed['partition'] == 'validation_0']['dataset_origin'].value_counts()
print(val_0_counts)

print("\nComposición de la partición 'training_0':")
train_0_counts = final_df_mixed[final_df_mixed['partition'] == 'training_0']['dataset_origin'].value_counts()
print(train_0_counts)
```

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split, StratifiedGroupKFold
import os

# ==============================================================================
# 1. CONFIGURACIÓN PRINCIPAL
# ==============================================================================

# Archivo de duplicados (Ajusta la ruta si es necesario)


OUTPUT_PATH = "./partitions_trainSN_valSN_synthetic_dup.parquet" # Nombre específico para esta prueba

# Nombres de columnas
REAL_OID_COL = 'oid'
REAL_CANDID_COL = 'measurement_id'
SYNTHETIC_GAL_ID_COL = 'gal_objectId'
SYNTHETIC_DIFF_ID_COL = 'diff_sourceId'

FINAL_OID_COL = 'oid'
FINAL_CANDID_COL = 'measurement_id'
CLASS_COL = 'class'
GROUP_COL = 'group_id'
BAND_COL = 'band'
STRATIFY_COL = 'stratify_col'

DUP_OID_COL = 'oid'
DUP_OID_KEPT_COL = 'oid_kept'

N_FOLDS = 5
RANDOM_STATE = 42
TEST_SIZE_NON_SN = 0.2 # Porcentaje de clases NO-SN para Test
```

```python
# ==============================================================================
# 3. ESTANDARIZACIÓN Y MANEJO DE DUPLICADOS (INTEGRADO)
# ==============================================================================
print("Estandarizando y procesando duplicados...")

# --- Datos Reales ---
df_real.rename(columns={REAL_CANDID_COL: FINAL_CANDID_COL}, inplace=True)
df_real[FINAL_OID_COL] = df_real[REAL_OID_COL].astype(str)
df_real[FINAL_CANDID_COL] = df_real[FINAL_CANDID_COL].astype(str)
df_real['dataset_origin'] = 'real'

# --- LÓGICA DE DUPLICADOS (COMBINADA) ---
# 1. Cargar ambos archivos de duplicados (si existen)
dups_frames = []
if os.path.exists(DUPLICATES_CROSSMATCH_PATH):
    dups_frames.append(pd.read_csv(DUPLICATES_CROSSMATCH_PATH)[[DUP_OID_COL, DUP_OID_KEPT_COL]])
    print(DUPLICATES_CROSSMATCH_PATH)
if os.path.exists(DUPLICATES_ANGULAR_PATH):
    dups_frames.append(pd.read_csv(DUPLICATES_ANGULAR_PATH)[[DUP_OID_COL, DUP_OID_KEPT_COL]])
    print(DUPLICATES_ANGULAR_PATH)

if dups_frames:
    df_all_dups = pd.concat(dups_frames, ignore_index=True)
    # Crear mapa unificado: oid -> oid_kept
    dup_map = dict(zip(df_all_dups[DUP_OID_COL].astype(str), df_all_dups[DUP_OID_KEPT_COL].astype(str)))
    print(f"Total de relaciones de duplicados cargadas: {len(dup_map)}")
else:
    dup_map = {}
    print("No se encontraron archivos de duplicados. Usando mapeo identidad.")

# 2. Aplicar el mapa para crear el group_id
#    Si un oid es un duplicado, se reemplaza por su 'oid_kept'. Si no, se mantiene su propio oid.
df_real[GROUP_COL] = df_real[FINAL_OID_COL].map(dup_map).fillna(df_real[FINAL_OID_COL])


# --- Datos Sintéticos ---
df_sinteticos[CLASS_COL] = 'SN'
df_sinteticos.rename(columns={SYNTHETIC_GAL_ID_COL: FINAL_OID_COL}, inplace=True)
df_sinteticos[FINAL_OID_COL] = df_sinteticos[FINAL_OID_COL].astype(str)

# Candid único para sintéticos
if 'original_path' in df_sinteticos.columns:
    df_sinteticos[FINAL_CANDID_COL] = df_sinteticos['original_path'].apply(lambda x: os.path.splitext(os.path.basename(x))[0]).astype(str)
else:
    df_sinteticos[FINAL_CANDID_COL] = [f'FAKE_CANDID_{i}' for i in range(len(df_sinteticos))]

# Group ID para sintéticos: Usamos la galaxia anfitriona (oid)
# NOTA: Aquí asumimos que la "inyección" es el factor de agrupación principal.
# Si tienes diff_sourceId disponible, podrías concatenarlo: oid + "_" + diff_sourceId
df_sinteticos[GROUP_COL] = df_sinteticos[FINAL_OID_COL]
df_sinteticos['dataset_origin'] = 'synthetic'

# --- Estratificación ---
df_real['stratify_col'] = df_real[CLASS_COL].astype(str) + '_' + df_real[BAND_COL].astype(str)
df_sinteticos['stratify_col'] = df_sinteticos[CLASS_COL].astype(str) + '_' + df_sinteticos[BAND_COL].astype(str)
STRATIFY_COL = 'stratify_col'

print("Estandarización completada.")
```

```python
# ==============================================================================
# 4. LÓGICA DE TEST (Prueba de Concepto con Duplicados Resueltos)
# ==============================================================================
print("\n--- Construyendo Partición de Prueba de Concepto ---")

# 1. Separar SN reales de las otras clases
df_real_sn = df_real[df_real[CLASS_COL] == 'SN'].copy()
df_real_non_sn = df_real[df_real[CLASS_COL] != 'SN'].copy()

# 2. Dividir clases NO-SN (20% Test, 80% Train/Val)
#    IMPORTANTE: El split se hace sobre los GRUPOS ÚNICOS (que ya manejan los duplicados)
unique_non_sn_groups = df_real_non_sn[GROUP_COL].unique()

# Estratificación a nivel de grupo (proxy usando el primer valor del grupo)
group_labels = df_real_non_sn.groupby(GROUP_COL)[CLASS_COL].first() # Usamos solo CLASE para ser robustos

try:
    groups_train_val_non_sn, groups_test_non_sn = train_test_split(
        unique_non_sn_groups,
        test_size=TEST_SIZE_NON_SN,
        random_state=RANDOM_STATE,
        stratify=group_labels.loc[unique_non_sn_groups]
    )
except ValueError:
    print("Advertencia: Falló estratificación en split inicial. Usando aleatorio.")
    groups_train_val_non_sn, groups_test_non_sn = train_test_split(
        unique_non_sn_groups, test_size=TEST_SIZE_NON_SN, random_state=RANDOM_STATE
    )

# Filtrar DataFrames usando la columna GROUP_COL
df_non_sn_train_val = df_real_non_sn[df_real_non_sn[GROUP_COL].isin(groups_train_val_non_sn)].copy()
df_non_sn_test = df_real_non_sn[df_real_non_sn[GROUP_COL].isin(groups_test_non_sn)].copy()

# 3. Construir TEST SET
#    Incluye TODAS las SN reales (y sus duplicados, si los hubiera, gracias a que no las filtramos)
#    + 20% de los grupos NO-SN (con sus duplicados)
df_test = pd.concat([df_real_sn, df_non_sn_test], ignore_index=True)
df_test['partition'] = 'test'

print(f"Test Set: {len(df_test)} filas. (Incluye {len(df_real_sn)} filas de SN reales)")

# 4. Construir TRAIN/VAL POOL
#    Incluye TODAS las SN Sintéticas
#    + 80% de los grupos NO-SN (con sus duplicados)
df_train_val_pool = pd.concat([df_sinteticos, df_non_sn_train_val], ignore_index=True).reset_index(drop=True)

print(f"Train/Val Pool: {len(df_train_val_pool)} filas. (SN son sintéticas)")
```

```python
# ==============================================================================
# 5. K-FOLD (Train/Val Sintético para SN)
# ==============================================================================
cols_to_save = [FINAL_OID_COL, FINAL_CANDID_COL, CLASS_COL, 'partition', 'dataset_origin']
partitions_list = [df_test[cols_to_save]]

X = df_train_val_pool.index.to_numpy()
y = df_train_val_pool[STRATIFY_COL].to_numpy()
groups = df_train_val_pool[GROUP_COL].to_numpy()

sgkf = StratifiedGroupKFold(n_splits=N_FOLDS, shuffle=True, random_state=RANDOM_STATE)

print(f"\nGenerando {N_FOLDS} folds con StratifiedGroupKFold...")

# Intentar estratificación compuesta, fallback a simple
try:
    splitter = sgkf.split(X, y, groups=groups)
    next(splitter); splitter = sgkf.split(X, y, groups=groups) 
except ValueError:
    y = df_train_val_pool[CLASS_COL].to_numpy()
    splitter = sgkf.split(X, y, groups=groups)

for fold_idx, (train_idx, val_idx) in enumerate(splitter):
    train_df = df_train_val_pool.iloc[train_idx].copy()
    val_df = df_train_val_pool.iloc[val_idx].copy()
    
    train_df['partition'] = f'training_{fold_idx}'
    val_df['partition'] = f'validation_{fold_idx}'
    
    partitions_list.extend([train_df[cols_to_save], val_df[cols_to_save]])

final_df = pd.concat(partitions_list, ignore_index=True)
```

```python
import pandas as pd
import numpy as np
from astropy.coordinates import SkyCoord, match_coordinates_sky
from astropy import units as u
import itertools

df_analysis = pd.merge(partitions_poc, df_real, on='oid', how='inner')
# Si partitions tiene filas por detección, nos quedamos con una por objeto
df_analysis = df_analysis.drop_duplicates(subset=['oid']) 

# ==============================================================================
# 1. CARGAR DATOS Y PREPARAR
# ==============================================================================
# ... (Asumiendo que df_analysis ya tiene 'oid', 'ra', 'dec', 'partition' cargados como antes) ...

print(f"Total de objetos únicos para analizar: {len(df_analysis)}")

# Función auxiliar para obtener coords de una partición
def get_coords(df, partition_name):
    subset = df[df['partition'] == partition_name]
    coords = SkyCoord(ra=subset['ra'].astype(float).to_numpy()*u.degree, 
                      dec=subset['dec'].astype(float).to_numpy()*u.degree)
    return coords, subset['oid'].to_numpy()

# ==============================================================================
# 2. CÁLCULO DE SEPARACIÓN ENTRE CONJUNTOS
# ==============================================================================
print("\n--- Análisis de Separación ENTRE Conjuntos (Busca fugas) ---")

# Definir los pares a comparar. 
# Queremos comparar cada fold de entrenamiento con su validación, y todo con test.
# Pero para simplificar y ver lo peor, compararemos todas las combinaciones posibles de particiones.

partition_names = df_analysis['partition'].unique()
# Generar todas las combinaciones de 2 particiones distintas
pairs = list(itertools.combinations(partition_names, 2))

resultados_cross = []

for part1, part2 in pairs:
    # Optimización: Solo nos interesa comparar Train vs Val del MISMO fold, o Train vs Test.
    # No tiene sentido comparar Training_0 con Validation_1, por ejemplo.
    
    es_train_val_mismo_fold = (part1.startswith('training_') and part2.replace('validation', 'training') == part1) or \
                              (part2.startswith('training_') and part1.replace('validation', 'training') == part2)
    
    es_con_test = ('test' in part1) or ('test' in part2)
    
    # Solo procesamos si es una comparación relevante para fuga de datos
    if es_train_val_mismo_fold or es_con_test:
        
        print(f"Comparando {part1} vs {part2}...")
        
        coords1, oids1 = get_coords(df_analysis, part1)
        coords2, oids2 = get_coords(df_analysis, part2)
        
        if len(coords1) == 0 or len(coords2) == 0:
            continue
            
        # match_coordinates_sky encuentra para cada objeto en coords1, el más cercano en coords2
        idx, d2d, _ = match_coordinates_sky(coords1, coords2)
        
        # Encontrar la mínima distancia en este cruce
        min_dist_idx = np.argmin(d2d)
        min_dist_arcsec = d2d[min_dist_idx].arcsec
        
        resultados_cross.append({
            'set_1': part1,
            'set_2': part2,
            'min_separation_arcsec': min_dist_arcsec,
            'oid_1': oids1[min_dist_idx],
            'oid_2': oids2[idx[min_dist_idx]]
        })

# ==============================================================================
# 3. RESULTADOS
# ==============================================================================
if resultados_cross:
    df_cross_results = pd.DataFrame(resultados_cross).sort_values('min_separation_arcsec')
    
    print("\n" + "="*80)
    print("RESUMEN DE SEPARACIÓN MÍNIMA ENTRE CONJUNTOS (Train vs Val / Train vs Test):")
    print(df_cross_results)
    print("="*80)
    
    # Alerta de Fuga
    limite_critico = 1.0 # arcsec (Si es menor a esto, probablemente es el mismo objeto)
    fugas = df_cross_results[df_cross_results['min_separation_arcsec'] < limite_critico]
    
    if not fugas.empty:
        print(f"\n🚨 ALERTA ROJA: Se encontraron {len(fugas)} pares de conjuntos con objetos a < {limite_critico}\".")
        print("Esto indica FUGA DE DATOS (Data Leakage). El mismo objeto físico está en ambos sets con distinto ID.")
        print(fugas)
    else:
        print(f"\n✅ EXCELENTE: La separación mínima entre conjuntos es > {limite_critico}\". No hay evidencia espacial de fuga de datos.")
```

```python
import pandas as pd
import numpy as np

def get_fold_data_in_ram(partitions_path, df_real_stamps, df_synth_stamps, fold_idx, partition_type='training'):
    """
    Obtiene los arrays X (imágenes) e y (etiquetas) para un fold específico,
    buscando eficientemente en el dataframe correcto.
    
    partitions_path: Ruta al archivo parquet creado.
    df_real_stamps: DataFrame con estampillas reales cargadas en RAM.
    df_synth_stamps: DataFrame con estampillas sintéticas cargadas en RAM.
    fold_idx: Número del fold (0, 1, 2, 3, 4).
    partition_type: 'training' o 'validation'.
    """
    
    print(f"--- Cargando {partition_type} para Fold {fold_idx} ---")
    
    # 1. Cargar particiones y filtrar por el fold deseado
    partitions = pd.read_parquet(partitions_path)
    target_partition = f"{partition_type}_{fold_idx}"
    subset_ids = partitions[partitions['partition'] == target_partition].copy()
    
    # 2. Separar los IDs según su origen usando la nueva columna
    real_ids = subset_ids[subset_ids['dataset_origin'] == 'real']['measurement_id']
    synth_ids = subset_ids[subset_ids['dataset_origin'] == 'synthetic']['measurement_id']
    
    print(f"  - Buscando {len(real_ids)} muestras reales...")
    print(f"  - Buscando {len(synth_ids)} muestras sintéticas...")
    
    # 3. Hacer el merge con el DataFrame correspondiente
    #    Aseguramos que 'measurement_id' sea string en los DF de origen para el merge
    df_real_stamps['measurement_id'] = df_real_stamps['measurement_id'].astype(str)
    df_synth_stamps['measurement_id'] = df_synth_stamps['measurement_id'].astype(str)

    # Merge Reales
    data_real = pd.merge(
        df_real_stamps, 
        real_ids.to_frame(name='measurement_id'), 
        on='measurement_id', 
        how='inner'
    )
    
    # Merge Sintéticos
    data_synth = pd.merge(
        df_synth_stamps, 
        synth_ids.to_frame(name='measurement_id'), 
        on='measurement_id', 
        how='inner'
    )
    
    # 4. Concatenar y mezclar (shuffle)
    full_data = pd.concat([data_real, data_synth], ignore_index=True)
    full_data = full_data.sample(frac=1, random_state=42).reset_index(drop=True) # Shuffle importante para training
    
    # 5. Extraer arrays (Asumiendo columna 'stamp' y columna 'class')
    # Nota: np.stack convierte una lista de arrays (21,21) en un tensor (N, 21, 21)
    X = np.stack(full_data['stamp'].values)
    y = full_data['class'].values # Aquí deberías aplicar tu encoding (one-hot o LabelEncoder)
    
    print(f"  -> Total cargado: {len(X)} imágenes. Shape: {X.shape}")
    
    return X, y

# --- EJEMPLO DE USO ---

# Supongamos que ya cargaste tus imagenes en RAM en estos dos DFs
# df_real_loaded = ...
# df_synth_loaded = ...

# X_train, y_train = get_fold_data_in_ram(
#     "partitions_final.parquet", 
#     df_real_loaded, 
#     df_synth_loaded, 
#     fold_idx=0, 
#     partition_type='training'
# )

# X_val, y_val = get_fold_data_in_ram(
#     "partitions_final.parquet", 
#     df_real_loaded, 
#     df_synth_loaded, 
#     fold_idx=0, 
#     partition_type='validation'
# )
```

```python
# Celda 4: Particionado

# Paso 1: Definir qué va al Test Set (TODAS las SN reales + muestra de no-SN reales)
real_sn_candids = df_real[df_real[CLASS_COL] == 'SN'][CANDID_COL].unique()
real_non_sn = df_real[df_real[CLASS_COL] != 'SN']

_, non_sn_for_test = train_test_split(
    real_non_sn, test_size=TEST_SIZE_NON_SN, random_state=RANDOM_STATE, stratify=real_non_sn[STRATIFY_COL]
)
test_candids = np.concatenate([real_sn_candids, non_sn_for_test[CANDID_COL].unique()])

# Paso 2: Dividir el DataFrame completo en Test y Train/Val
df_test = df_full[df_full[CANDID_COL].isin(test_candids)].copy()
df_train_val_pool = df_full[~df_full[CANDID_COL].isin(test_candids)].copy()
df_test['partition'] = 'test'

# Paso 3: Aplicar StratifiedGroupKFold al pool de Train/Val
partitions = [df_test[[OID_COL, CANDID_COL, CLASS_COL, 'partition']]]
groups = df_train_val_pool[GROUP_COL]
sgkf = StratifiedGroupKFold(n_splits=N_FOLDS, shuffle=True, random_state=RANDOM_STATE)

print(f"\nGenerando {N_FOLDS} folds...")
df_train_val_pool['fold'] = -1
for fold_idx, (_, val_idx) in enumerate(sgkf.split(df_train_val_pool, df_train_val_pool[STRATIFY_COL], groups=groups)):
    df_train_val_pool.iloc[val_idx, df_train_val_pool.columns.get_loc('fold')] = fold_idx

for fold_idx in range(N_FOLDS):
    train_df = df_train_val_pool[df_train_val_pool['fold'] != fold_idx].copy()
    val_df = df_train_val_pool[df_train_val_pool['fold'] == fold_idx].copy()
    train_df['partition'] = f'training_{fold_idx}'
    val_df['partition'] = f'validation_{fold_idx}'
    partitions.extend([
        train_df[[OID_COL, CANDID_COL, CLASS_COL, 'partition']],
        val_df[[OID_COL, CANDID_COL, CLASS_COL, 'partition']]
    ])

final_df_partitions = pd.concat(partitions, ignore_index=True)

# GUARDAR
os.makedirs(os.path.dirname(output_path), exist_ok=True)
final_df_partitions.to_parquet(output_path, index=False)
print(f"\nArchivo de particiones guardado en: {output_path}")
print("\nResumen de particiones creadas:\n", final_df_partitions['partition'].value_counts().sort_index())
```

```python
def create_partitions(df_to_partition, group_col, stratify_col, n_folds, random_state):
    """
    Toma un DataFrame y devuelve un DataFrame con una nueva columna 'fold'
    indicando a qué fold de validación cruzada pertenece cada fila.
    Respeta los grupos para que no se mezclen entre train y val.
    """
    df_partitioned = df_to_partition.copy()
    df_partitioned['fold'] = -1

    groups = df_partitioned[group_col].fillna(df_partitioned[OID_COL])
    
    sgkf = StratifiedGroupKFold(n_splits=n_folds, shuffle=True, random_state=random_state)
    
    for fold_idx, (_, val_idx) in enumerate(sgkf.split(df_partitioned, df_partitioned[stratify_col], groups=groups)):
        df_partitioned.loc[df_partitioned.index[val_idx], 'fold'] = fold_idx
        
    print(f"Se crearon {n_folds} folds. Distribución de filas por fold:")
    print(df_partitioned['fold'].value_counts())
    
    return df_partitioned
```

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split, StratifiedGroupKFold
import os
```

```python
# --- 5. EJECUCIÓN DE LA PRUEBA DE CONCEPTO ---
print("--- Ejecutando Prueba de Concepto ---")

# a. El conjunto de TEST son TODOS los datos reales
df_test_poc = df_real.copy()
df_test_poc['partition'] = 'test'

# b. El conjunto a particionar en TRAIN/VAL son las SN sintéticas + clases reales no-SN
df_reales_negativos = df_real[df_real[CLASS_COL] != 'SN'].copy()
df_train_val_poc = pd.concat([df_sinteticos, df_reales_negativos], ignore_index=True)

# c. Aplicar la partición K-Fold
df_partitioned_poc = create_partitions(
    df_train_val_poc,
    group_col=GROUP_COL,
    stratify_col=STRATIFY_COL,
    n_folds=N_FOLDS,
    random_state=RANDOM_STATE
)

# d. Construir el DataFrame final con la columna 'partition'
partitions_poc = [df_test_poc]
for fold_idx in range(N_FOLDS):
    train_df = df_partitioned_poc[df_partitioned_poc['fold'] != fold_idx].copy()
    val_df = df_partitioned_poc[df_partitioned_poc['fold'] == fold_idx].copy()
    
    train_df['partition'] = f'training_{fold_idx}'
    val_df['partition'] = f'validation_{fold_idx}'
    
    partitions_poc.extend([train_df, val_df])

final_df_poc = pd.concat(partitions_poc, ignore_index=True)

# e. Guardar y verificar
output_path_poc = "./partitions_proof_of_concept.parquet"
# Columnas a mantener
cols_to_keep = [OID_COL, CANDID_COL, CLASS_COL, 'partition']
final_df_poc = final_df_poc[cols_to_keep]
os.makedirs(os.path.dirname(output_path_poc), exist_ok=True)
final_df_poc.to_parquet(output_path_poc, index=False)

print(f"\nDataFrame de 'Prueba de Concepto' guardado en {output_path_poc}")
print("Resumen de particiones creadas:\n", final_df_poc['partition'].value_counts())
```

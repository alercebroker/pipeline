```python
PATH_STAMPS_FILES = './data/processed/ts_stamps_v0.0.2_comm_4candmax/stamps/*.pkl'
PATH_STAMPS_OBJECTS = './data/processed/ts_stamps_v0.0.2_comm_4candmax/comm_objs_4candmax.parquet'

DUP_TNS_PATH  = "./data/processed/ts_stamps_v0.0.2_comm_4candmax/comm_agn_sne_vs_tran_dup_remove.csv"
DUP_ANGULAR_DISTANCE_PATH  = "/home/dmoreno/pipeline_v4_final/pipeline/training/stamp_classifier/models/experimentation/stamp_rubin/notebooks/oids2remove.csv"

SCHEMA_BBDD = 'multisurvey'
LSST_sid = 1
```

```python
import json

with open('credentials.json', 'r', encoding='utf-8') as archivo:
    params = json.load(archivo)
    
# Necesitamos conectarnos para mapear los nombres de las bandas

import sqlalchemy as sa
from sqlalchemy import inspect
from sqlalchemy import text

import pandas as pd

engine = sa.create_engine('postgresql+psycopg2://' + params['user'] \
                          + ':' + params['password'] + '@' + params['host'] \
                          + '/' + params['dbname'])

conn = engine.connect()

inspector = inspect(engine)
tables = inspector.get_table_names()
print('Available Tables:\n', tables)

schemas = inspector.get_schema_names()

print("\nSchemas disponibles en la base de datos:")
print(schemas)

tables_rubin_schema = inspector.get_table_names(schema=SCHEMA_BBDD)

query = f'''
SELECT
    *
FROM
    {SCHEMA_BBDD}.bands
'''
df_bands = pd.read_sql_query(text(query), conn)
dict_mapping = df_bands[df_bands.sid == LSST_sid].set_index('band')['band_name'].to_dict()
dict_mapping
```

```python
# Alejandra dejo este archivo
df_duplicates_tns = pd.read_csv(DUP_TNS_PATH)
df_duplicates_tns
```

```python
# Francisco Forster dejo este archivo
df_duplicates_ang = pd.read_csv(DUP_ANGULAR_DISTANCE_PATH)
df_duplicates_ang
```

```python
mapping_oids = dict(zip(df_duplicates_tns['oid'], df_duplicates_tns['oid_kept']))
mapping_oids
```

```python
# Desde aquí en adelante empezamos a utilizar solo una observación por stamp

stamps_objs_cp = stamps_objs.copy()
stamps_objs_cp = stamps_objs_cp.groupby('oid').first().reset_index()
stamps_objs_cp.band = stamps_objs_cp.band.map(dict_mapping)
stamps_objs_cp['oid_kept'] = stamps_objs_cp['oid']
stamps_objs_cp = stamps_objs_cp[['oid_kept', 'oid', 'measurement_id', 'band', 'class', 'ra', 'dec', 'mjd']]
stamps_objs_cp = stamps_objs_cp.sort_values(['oid_kept', 'mjd'])
stamps_objs_cp['dataset_origin'] = 'real'
stamps_objs_cp['synth_file_path'] = np.nan

df_stamps_clean_tns = stamps_objs_cp.copy()
df_stamps_clean_tns.oid_kept = df_stamps_clean_tns.oid_kept.replace(mapping_oids)
df_stamps_clean_tns = df_stamps_clean_tns.groupby('oid_kept').first().reset_index()
df_stamps_clean_tns
```

```python
df_dup_stamps_ang = df_stamps_clean_tns[
    (df_stamps_clean_tns.oid.isin(df_duplicates_ang.oid))
    ]

df_stamps_clean_tns_ang = df_stamps_clean_tns[
    (~df_stamps_clean_tns.oid.isin(df_duplicates_ang.oid))
    ]

print('stamps_objs_cp', stamps_objs_cp.shape)
print('df_stamps_clean_tns', df_stamps_clean_tns.shape)
print('df_dup_stamps_ang', df_dup_stamps_ang.shape)
print('df_stamps_clean_tns_ang', df_stamps_clean_tns_ang.shape)
```

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import StratifiedKFold

unique_objects_df = df_stamps_clean_tns_ang[['oid_kept', 'class']].drop_duplicates().reset_index(drop=True)

# 2. División en Test y Train/Val
test_oids = []
trainval_oids = []
for class_name in unique_objects_df['class'].unique():
    class_oids_df = unique_objects_df[unique_objects_df['class'] == class_name]
    shuffled_oids = class_oids_df['oid_kept'].sample(frac=1, random_state=42)
    test_frac = 0.60 if class_name == 'SN' else 0.20
    split_idx = int(len(shuffled_oids) * test_frac)
    test_oids.extend(shuffled_oids.iloc[:split_idx])
    trainval_oids.extend(shuffled_oids.iloc[split_idx:])
    
test_objects_df = unique_objects_df[unique_objects_df['oid_kept'].isin(test_oids)]
trainval_objects_df = unique_objects_df[unique_objects_df['oid_kept'].isin(trainval_oids)]

print('test_objects_df:\n', test_objects_df['class'].value_counts())
print('\ntrainval_objects_df:\n', trainval_objects_df['class'].value_counts())
```

```python
# 3. Crear los 5 Folds con la lógica personalizada
N_SPLITS = 5
all_partitions = []

df_test_partitions = test_objects_df.copy()
df_test_partitions['partition'] = 'test'
all_partitions.append(df_test_partitions)

sn_trainval_df = trainval_objects_df[trainval_objects_df['class'] == 'SN']
other_trainval_df = trainval_objects_df[trainval_objects_df['class'] != 'SN']

if not other_trainval_df.empty:
    skf = StratifiedKFold(n_splits=N_SPLITS, shuffle=True, random_state=42)
    other_trainval_df['fold'] = -1
    for fold_num, (train_idx, val_idx) in enumerate(skf.split(other_trainval_df, other_trainval_df['class'])):
        val_oids = other_trainval_df.iloc[val_idx]['oid_kept']
        other_trainval_df.loc[other_trainval_df['oid_kept'].isin(val_oids), 'fold'] = fold_num

for i in range(N_SPLITS):
    if not other_trainval_df.empty:
        df_other_perspective = other_trainval_df.copy()
        df_other_perspective['partition'] = np.where(df_other_perspective['fold'] == i, f'validation_{i}', f'training_{i}')
        all_partitions.append(df_other_perspective[['oid_kept', 'class', 'partition']])

    if not sn_trainval_df.empty:
        validation_sn_oids = sn_trainval_df['oid_kept'].sample(frac=0.6, random_state=42 + i)
        df_sn_perspective = sn_trainval_df.copy()
        df_sn_perspective['partition'] = np.where(df_sn_perspective['oid_kept'].isin(validation_sn_oids), f'validation_{i}', f'training_{i}')
        all_partitions.append(df_sn_perspective[['oid_kept', 'class', 'partition']])

main_mapping_df = pd.concat(all_partitions, ignore_index=True)

# Unir el mapeo con el DataFrame original principal
final_partition_df_main = pd.merge(
    df_stamps_clean_tns_ang,
    main_mapping_df,
    on=['oid_kept', 'class'],
    how='left'
)

# --- PASO B: Procesar el DataFrame adicional (df_dup_stamps_ang) ---
print("\n--- Procesando datos adicionales para entrenamiento ---")

extra_training_partitions = []
# Replicamos los datos para cada uno de los 5 folds, asignándolos siempre a 'training'
for i in range(N_SPLITS):
    df_fold_extra = df_dup_stamps_ang.copy()
    df_fold_extra['partition'] = f'training_{i}'
    extra_training_partitions.append(df_fold_extra)

# Concatenamos las 5 réplicas en un único DataFrame
df_extra_partitioned = pd.concat(extra_training_partitions, ignore_index=True)

print(f"Se han generado {len(df_extra_partitioned)} filas a partir de los datos adicionales.")
print(f"({len(df_dup_stamps_ang)} filas originales x {N_SPLITS} folds)")
print("-" * 50)


# --- PASO C: Combinar ambos resultados ---

final_partition_df = pd.concat(
    [final_partition_df_main, df_extra_partitioned],
    ignore_index=True
)


# --- 4. Verificación y Formato de Salida ---
print("\n--- Verificación Final (Fold 0) ---")
# Tomamos un objeto del set adicional para verificar
oid_to_check = df_dup_stamps_ang['oid_kept'].iloc[0]
print(f"Verificando particiones para el objeto de muestra '{oid_to_check}':")
print(final_partition_df[final_partition_df['oid_kept'] == oid_to_check][['oid_kept', 'partition']])

print("\nConteo de objetos únicos en las particiones del Fold 0:")
check_df = final_partition_df[['oid_kept', 'class', 'partition']].drop_duplicates()
fold_0_counts = check_df[check_df['partition'].isin(['training_0', 'validation_0'])]
print(fold_0_counts.groupby(['partition', 'class']).size().unstack(fill_value=0))
print("-" * 50)


# Reordenar las columnas para el formato final
output_columns = [
    'oid', 'oid_kept', 'measurement_id', 'band', 'ra', 'dec', 'mjd', 
    'class', 'partition', 'dataset_origin', 'synth_file_path'
]
final_partition_df = final_partition_df[output_columns]
```

```python
objs = stamps_objs.copy()[['oid', 'measurement_id', 'band', 'class', 'ra', 'dec', 'mjd']]

objs.band = objs.band.map(dict_mapping)
objs['oid_kept'] = objs.oid
objs.oid_kept = objs.oid_kept.replace(mapping_oids)
objs['dataset_origin'] = 'real'
objs['synth_file_path'] = np.nan
objs
```

```python
partition_map = final_partition_df[['oid_kept', 'class', 'partition']].drop_duplicates()

print("\n--- Mapa de Partición Limpio (lo que se usará para el merge) ---")
print(partition_map)
print(f"\nForma: {partition_map.shape}")
print("-" * 50)


# --- Paso 2: Unir (Merge) los datos completos con el mapa de partición ---
# Usamos un 'left' merge: empezamos con `objs` (todos los datos) y le añadimos
# la columna 'partition' del mapa, coincidiendo por `oid_kept` y `class`.
# De esta forma, cada observación en `objs` recibirá la(s) etiqueta(s) de partición
# que le corresponde a su objeto.

final_dataset_partitioned = pd.merge(
    objs,
    partition_map,
    on=['oid_kept', 'class'], # Columnas clave para la unión
    how='left'
).reset_index(drop=True)

final_dataset_partitioned
```

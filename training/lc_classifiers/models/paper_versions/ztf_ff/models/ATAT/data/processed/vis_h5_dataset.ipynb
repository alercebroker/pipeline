```python
import h5py
import numpy as np

# Abrir archivo original
input_path = './ds_pre250408_ndetge8_sanchez_tax_20folds_pos260202/dataset.h5'
output_path = './dataset_window_0.h5'

with h5py.File(input_path, 'r') as h5_in:
    # Identificar índices que terminan en '_0'
    snids = h5_in['SNID'][:]
    mask_window0 = np.array([s.decode().endswith('_0') for s in snids])
    indices = np.where(mask_window0)[0]
    
    print(f"Seleccionados {len(indices)} objetos con ventana 0.")

    # Crear nuevo archivo .h5 con solo esos índices
    with h5py.File(output_path, 'w') as h5_out:
        for key in h5_in.keys():
            data = h5_in[key]
            if data.shape[0] == snids.shape[0]:
                # Dataset con la misma cantidad de objetos
                h5_out.create_dataset(key, data=data[indices])
            else:
                # Otros datasets como 'training_x', 'validation_x', etc., se omiten o copian completos si quieres
                print(f"Saltando key: {key} (shape: {data.shape})")
        
print("Nuevo archivo guardado en:", output_path)
```

```python
import h5py
import numpy as np

input_path = './ds_pre250408_ndetge8_20folds_pos250514_float32/dataset.h5'
output_path = './dataset_window0.h5'

with h5py.File(input_path, 'r') as h5_in:
    # ⚡ Convertimos todos los SNID (bytes) a str de forma segura
    snid_raw = h5_in['SNID'][:]
    snids = np.array([s.decode('utf-8') for s in snid_raw])

    # Vectorizado: endswith '_0'
    mask_window0 = np.char.endswith(snids, '_0')
    indices = np.where(mask_window0)[0]

    print(f"✅ Seleccionados {len(indices)} objetos con ventana _0")

    # Creamos el nuevo archivo
    with h5py.File(output_path, 'w') as h5_out:
        for key in h5_in.keys():
            data = h5_in[key]
            if data.shape[0] == snid_raw.shape[0]:
                print(f"📦 Copiando: {key}")
                h5_out.create_dataset(key, data=data[indices], compression='gzip')
            else:
                print(f"⏭️ Saltando: {key} (shape: {data.shape})")

print("✅ Nuevo archivo guardado en:", output_path)
```

```python
import h5py
import numpy as np
import pandas as pd

file_path = './ds_pre250408_ndetge8_20folds_pos250514_float32/dataset.h5'
h5_file = h5py.File(file_path, 'r')

# Decodificar SNIDs y filtrar por '_0'
snid = np.array([s.decode('utf-8') for s in h5_file['SNID'][:]])
mask_window0 = np.char.endswith(snid, '_0')
indices = np.where(mask_window0)[0]
print(f"Número de objetos con _0: {len(indices)}")

# Construir el all_data estilo create_dataset_h5py
all_data = {
    'oid': snid[indices],
    'time': h5_file['time'][indices],
    'brightness': h5_file['brightness'][indices],
    'e_brightness': h5_file['e_brightness'][indices],
    'mask': h5_file['mask'][indices],
    'time_detection': h5_file['time_detection'][indices],
    'mask_detection': h5_file['mask_detection'][indices],
    'time_photometry': h5_file['time_photometry'][indices],
    'mask_photometry': h5_file['mask_photometry'][indices],
    'label_int': h5_file['labels'][indices],
    'class_name': h5_file['class_name'][indices],
    # Si tienes metadata/features
    'metadata_feat': h5_file['metadata_feat'][indices],
}

for feat_name in h5_file.keys():
    if feat_name.startswith('extracted_feat_') and h5_file[feat_name].shape[0] == snid.shape[0]:
        all_data[feat_name] = h5_file[feat_name][indices]

output_path = './dataset_window0_fast.h5'
with h5py.File(output_path, 'w') as hf:
    hf.create_dataset("SNID", data=np.array(all_data["oid"], dtype='S'))  # string como bytes
    hf.create_dataset("time", data=all_data["time"].astype(np.float32))
    hf.create_dataset("brightness", data=all_data["brightness"].astype(np.float32))
    hf.create_dataset("e_brightness", data=all_data["e_brightness"].astype(np.float32))
    hf.create_dataset("mask", data=all_data["mask"].astype(bool))
    hf.create_dataset("time_detection", data=all_data["time_detection"].astype(np.float32))
    hf.create_dataset("mask_detection", data=all_data["mask_detection"].astype(bool))
    hf.create_dataset("time_photometry", data=all_data["time_photometry"].astype(np.float32))
    hf.create_dataset("mask_photometry", data=all_data["mask_photometry"].astype(bool))
    hf.create_dataset("labels", data=all_data["label_int"].astype(np.int8))
    hf.create_dataset("class_name", data=all_data["class_name"])  # tipo bytes

    # Agregar metadata y features si están
    if "metadata_feat" in all_data:
        hf.create_dataset("metadata_feat", data=all_data["metadata_feat"].astype(np.float32))

    for key in all_data:
        if key.startswith("extracted_feat_"):
            hf.create_dataset(key, data=all_data[key].astype(np.float32))

print("✅ Nuevo archivo rápido guardado:", output_path)
```

```python
import numpy as np
import pandas as pd

# Reorganizar el array para obtener flujos y bandas en formato 2D
flux = h5_file.get("extracted_feat_None")[:].reshape(-1, 2)  # Combinar las dos primeras dimensiones (flujos)
bands = np.tile([1, 2], flux.shape[0] // 2)  # Crear una "columna" de bandas (1 y 2)

# Crear una función para calcular estadísticas por grupo (banda)
def calculate_group_stats(flux, bands):
    unique_bands = np.unique(bands)
    stats = []
    for band in unique_bands:
        band_flux = flux[bands == band]
        stats.append({
            "fid": band,
            "Mean": round(np.mean(band_flux), 4),
            "Std": round(np.std(band_flux), 4),
            "Min": round(np.min(band_flux), 4),
            "P25": round(np.percentile(band_flux, 25), 4),
            "Median": round(np.median(band_flux), 4),
            "P75": round(np.percentile(band_flux, 75), 4),
            "Max": round(np.max(band_flux), 4),
        })
    return pd.DataFrame(stats)

# Calcular estadísticas
stats_df = calculate_group_stats(flux, bands)
print(stats_df)
```
